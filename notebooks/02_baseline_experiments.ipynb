# 02 — Baseline experiments

Reproduces the feasibility results in D1 section 6. The runner is in
`frd.experiments`; this notebook just calls it and looks at the output,
so the notebook and the command line cannot disagree.

```python
import sys; sys.path.insert(0, "../src")
import pandas as pd
from frd import experiments
```

## Run every model in every condition

```python
HUMAN, MACHINE = "synthetic_human", "synthetic_machine"
results = experiments.run_all(HUMAN, MACHINE, vectorisers=("word", "char"))
results.head()
```

## The headline table

Within-dataset against cross-dataset. The gap between these two columns
is the project.

```python
results.pivot_table(index=["model", "vectoriser"], columns="condition",
                   values="macro_f1", aggfunc="mean").round(3)
```

## Direction of transfer

The drop is not symmetric: training on one population and testing on the
other is harder in one direction than the other. Worth a sentence in the
discussion.

```python
cross = results[results["condition"] == "cross"]
cross.pivot_table(index="train_source", columns="test_source",
                  values="macro_f1", aggfunc="mean").round(3)
```

## Persist and plot

```python
experiments.append_results(results)

from frd import figures
figures.condition_comparison(pd.read_csv(experiments.RESULTS_FILE))
```
